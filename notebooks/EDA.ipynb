Model Pipeline

```python
#divide the data into input and output
X=df.drop('Amount_Spent',axis=1)
y=df.Amount_Spent
```

```python
#split the data into training and testing datasets
Xtrain,Xtest,ytrain,ytest = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
#separate numeric and object column names
num_cols = Xtrain.select_dtypes(include="number").columns
cat_cols = Xtrain.select_dtypes(include="str").columns
```

```python
#multi - models
models = {
    "Linear regression":Pipeline([("preprocessor",ColumnTransformer([("encoder",OneHotEncoder(),cat_cols),
                                             ("scaler",StandardScaler(),num_cols)])),
          ("model",LinearRegression())]),
    "Ridge regression":Pipeline([("preprocessor",ColumnTransformer([("encoder",OneHotEncoder(),cat_cols),
                                             ("scaler",StandardScaler(),num_cols)])),
          ("model",Ridge(alpha=50))]),
    "Lasso regression":Pipeline([("preprocessor",ColumnTransformer([("encoder",OneHotEncoder(),cat_cols),
                                             ("scaler",StandardScaler(),num_cols)])),
          ("model",Lasso(alpha=100))]),
    "Random forest regression":Pipeline([("preprocessor",ColumnTransformer([("encoder",OneHotEncoder(),cat_cols),
                                             ("scaler",StandardScaler(),num_cols)])),
          ("model",RandomForestRegressor(n_estimators=100, random_state=42))]),
    "Decision tree regression":Pipeline([("preprocessor",ColumnTransformer([("encoder",OneHotEncoder(),cat_cols),
                                             ("scaler",StandardScaler(),num_cols)])),
          ("model",DecisionTreeRegressor(max_depth=10, random_state=42))])

    
}
```

```python
#multi model training
model_pref={}
for model_name,pipeline in models.items():
    pipeline.fit(Xtrain,ytrain)
    ypred=pipeline.predict(Xtest)
    r2=r2_score(ytest,ypred)
    mse=mean_squared_error(ytest,ypred)
    mae=mean_absolute_error(ytest,ypred)
    model_pref[model_name]={'R2':r2,'MAE':mae,'MSE':mse}
```

```python
#model performance table
perf_df=pd.DataFrame(model_pref)
perf_df
```

```python
#fetch best model
best_model_data=max(model_pref.items(),key=lambda d:d[1]['R2'])
best_model_name=best_model_data[0]
print(best_model_data)
best_pipeline= models[best_model_name]
```

```python
#fit the best pipeline before making pred
best_pipeline.fit(Xtrain,ytrain)
best_predictions=best_pipeline.predict(Xtest)
```
